```python
# Recency scoring (lower is better)
rfm["R_Score"] = pd.qcut(rfm["Recency"].rank(method="first"),5,labels=[5, 4, 3, 2, 1])

# Frequency scoring (higher is better)
rfm["F_Score"] = pd.qcut(rfm["Frequency"].rank(method="first"),5,labels=[1, 2, 3, 4, 5])

# Monetary scoring (higher is better)
rfm["M_Score"] = pd.qcut(rfm["Monetary"].rank(method="first"),5,labels=[1, 2, 3, 4, 5])

print(rfm.head(10))
```

```python
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str) +
    rfm["F_Score"].astype(str) +
    rfm["M_Score"].astype(str)
)

print(rfm[["Recency", "Frequency", "Monetary","R_Score", "F_Score", "M_Score", "RFM_Score"]].head(10))
```

```python
rfm.describe()
```

```python
print(rfm["RFM_Score"].value_counts().head(15))
```

```python
def segment_customer(row):

    r = int(row["R_Score"])
    f = int(row["F_Score"])
    m = int(row["M_Score"])

    # Best customers
    if r >= 4 and f >= 4 and m >= 4:
        return "Champions"

    # Customers who buy frequently and spend well
    elif r >= 3 and f >= 4:
        return "Loyal Customers"

    # Recent customers who have good potential
    elif r >= 4 and f <= 3:
        return "Potential Loyalists"

    # Customers who used to be valuable but are becoming inactive
    elif r <= 2 and f >= 4:
        return "At Risk"

    # Customers with low engagement and low value
    elif r <= 2 and f <= 2 and m <= 2:
        return "Lost Customers"

    # Remaining customers
    else:
        return "Needs Attention"

rfm["Segment"] = rfm.apply(segment_customer, axis=1)
```

```python
def retention_strategy(segment):

    if segment == "Champions":
        return "Reward loyalty and encourage referrals"

    elif segment == "Loyal Customers":
        return "Offer loyalty rewards and cross-sell products"

    elif segment == "Potential Loyalists":
        return "Encourage second and repeat purchases"

    elif segment == "At Risk":
        return "Send personalized win-back offers"

    elif segment == "Needs Attention":
        return "Use targeted re-engagement campaigns"

    elif segment == "Lost Customers":
        return "Use low-cost reactivation campaigns"

rfm["Retention_Strategy"] = rfm["Segment"].apply(retention_strategy)

print(rfm[["Segment", "Retention_Strategy"]].head(10))
```

## K-Means Customer Segmentation

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
```

### Select existing RFM features

```python
X = rfm[["Recency", "Frequency", "Monetary"]].copy()
```

### Reduce skewness

```python
# Reduce skewness
X_log = np.log1p(X)
X_log.shape
```

### Standardize features

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_log)
print("Input shape:", X_scaled.shape)
print("Feature means:", X_scaled.mean(axis=0).round(2))
print("Feature standard deviations:", X_scaled.std(axis=0).round(2))
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
```

```python
wcss = []

for k in range(1, 11):
    kmeans = KMeans(n_clusters=k,
                    random_state=42,
                    n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
for _ in wcss:
    print(_)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(wcss, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels,
        sample_size=2000,
        random_state=42
    )

    silhouette_scores.append(score)

    print(f"K = {k}, Silhouette Score = {score:.4f}")
print("Best score : ",max(silhouette_scores))
```

### Train K-Means with K = 2

```python
kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

rfm["Cluster"] = kmeans.fit_predict(X_scaled)

print(rfm.head(10))
```

```python
print(rfm["Cluster"].value_counts())
```

### what each cluster represents

```python
cluster_profile = rfm.groupby("Cluster").agg(Customers=("Recency", "count"),
                                             Avg_Recency=("Recency", "mean"),
                                             Avg_Frequency=("Frequency", "mean"),
                                             Avg_Monetary=("Monetary", "mean")
                                            ).round(2)

print(cluster_profile)
```

```python
for k in [3, 4]:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    temp_rfm = rfm.copy()
    temp_rfm["Cluster_Test"] = labels

    profile = temp_rfm.groupby("Cluster_Test").agg(
        Customers=("Recency", "count"),
        Avg_Recency=("Recency", "mean"),
        Avg_Frequency=("Frequency", "mean"),
        Avg_Monetary=("Monetary", "mean")
    ).round(2)

    print(f"\n========== K = {k} ==========")
    print(profile)
```

### compare K=4 clusters with our RFM segments

```python
historical_rfm = historical_rfm.drop(columns=["LastPurchaseDate"])
```

```python
# 2. Create RFM scores
historical_rfm["R_Score"] = pd.qcut(
    historical_rfm["Recency"].rank(method="first"),5,labels=[5, 4, 3, 2, 1]).astype(int)

historical_rfm["F_Score"] = pd.qcut(
  historical_rfm["Frequency"].rank(method="first"),5,labels=[1, 2, 3, 4, 5]).astype(int)

historical_rfm["M_Score"] = pd.qcut(
    historical_rfm["Monetary"].rank(method="first"),5,labels=[1, 2, 3, 4, 5]).astype(int)
```

```python
# 3. Create customer segments
conditions = [
    (historical_rfm["R_Score"] >= 4) &
    (historical_rfm["F_Score"] >= 4) &
    (historical_rfm["M_Score"] >= 4),

    (historical_rfm["R_Score"] >= 3) &
    (historical_rfm["F_Score"] >= 4),

    (historical_rfm["R_Score"] >= 4) &
    (historical_rfm["F_Score"] >= 2),

    (historical_rfm["R_Score"] <= 2) &
    (historical_rfm["F_Score"] >= 3),

    (historical_rfm["R_Score"] <= 2) &
    (historical_rfm["F_Score"] <= 2)
]

choices = [
    "Champions",
    "Loyal Customers",
    "Potential Loyalists",
    "At Risk",
    "Lost Customers"
]

historical_rfm["Segment"] = np.select(conditions,choices,default="Needs Attention")
```

```python
# 4. Add historical segment to our ML results
retention_output = retention_output.drop(columns=["Segment", "Retention_Strategy"],errors="ignore")

retention_output = retention_output.join(historical_rfm[["Segment"]])

print(retention_output[["Customer ID","Frequency","Monetary","Recency","Inactivity_Probability","Segment"]].head(15))
```

```python
at_risk_valuable = retention_output[(retention_output["Segment"].isin(["Champions", "Loyal Customers"])) &
                                                                     (retention_output["Inactivity_Probability"] >= 0.3)
                                                                     ].sort_values(by="Inactivity_Probability",ascending=False)

print(at_risk_valuable[["Customer ID","Segment","Frequency","Monetary","Recency","Inactivity_Probability"]].head(15))
```

### Creating a segment-wise risk summary

```python
segment_risk = retention_output.groupby("Segment").agg(
                                                        Total_Customers=("Customer ID", "count"),
                                                        Predicted_Inactive=("Predicted_Status", "sum"),
                                                        Average_Inactivity_Probability=("Inactivity_Probability", "mean"),
                                                        Total_Historical_Spending=("Monetary", "sum")
                                                    )

segment_risk["Predicted_Inactive_Percentage"] = (segment_risk["Predicted_Inactive"] / segment_risk["Total_Customers"] * 100)

segment_risk = segment_risk.sort_values("Predicted_Inactive_Percentage",ascending=False)

print(segment_risk.round(2))
```
